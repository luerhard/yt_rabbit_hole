```python
import networkx as nx
import pandas as pd

import src
```

```python
path = src.PATH / "data/interim/networks/"

out_path = src.PATH / "data/interim/node_lists"

out_path.mkdir(exist_ok=True, parents=True)

graph_files = list(path.iterdir())
```

```python
for graph_file in graph_files:
    csv_file = graph_file.name.replace(".gml", ".csv")

    g = nx.read_gml(graph_file)

    nodes = []
    for node, data in g.nodes(data=True):
        if data["step"] == 0:
            row = dict()
            row["video_id"] = node
            row["title"] = data["title"]
            row["description"] = data["description"]
            row["date"] = data["date"]
            row["fav_count"] = data["view_count"]
            row["url"] = f"https://www.youtube.com/watch?v={node}"
            nodes.append(row)
    df = pd.DataFrame(nodes)

    df.to_csv(out_path / csv_file, index=False)
```
