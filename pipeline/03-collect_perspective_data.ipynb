```python
from pathlib import Path

import pandas as pd
from tqdm.auto import tqdm

from src.perspective.request import PerspectiveRequest
```

```python
in_folder = Path("/home/lukas/git/yt_rabbit_hole/data/interim/node_lists/")
out_folder = Path("/home/lukas/git/yt_rabbit_hole/data/interim/perspective_data/")
out_folder.mkdir(exist_ok=True, parents=True)
```

```python
in_files = list(in_folder.iterdir())
```

```python
def extract_scores(item):
    scores = item["attributeScores"]
    r = dict()
    for name, val_dict in scores.items():
        r[f"PERSPECTIVE_{name}"] = val_dict["summaryScore"]["value"]

    return r
```

```python
myattrs = [
    "TOXICITY",
    "INSULT",
    "THREAT",
    "IDENTITY_ATTACK",
    "PROFANITY",
    "SEXUALLY_EXPLICIT",
    "FLIRTATION",
    "INFLAMMATORY",
]

perspective = PerspectiveRequest(attrs=myattrs)
```

```python
r = perspective.analyze("Why some people STILL believe the Earth is flat", "en")
```

```python
for file in in_files:
    filename = file.name
    out_file = out_folder / filename

    if out_file.is_file():
        continue
    print("Running", filename)

    df = pd.read_csv(file)

    results = []

    for row in tqdm(df.itertuples()):
        return_val = perspective.analyze(row.title, lang="en")
        result = extract_scores(return_val)
        result["video_id"] = row.video_id
        results.append(result)

    out = pd.DataFrame(results)

    out.to_csv(out_file, index=False)
```
